```python
import matplotlib.pyplot as plt
```

```python
x=range(1,11)
y1=list(map(lambda x:x**2,x))
y2=list(map(lambda x:x**3,x))
```

```python
x
```

```python
y1
```

```python
y2
```

```python
#line plot
plt.plot(x,y1,color="red",marker="o",linestyle=":",label="sqr")
plt.plot(x,y2,color="orange",marker="*",linestyle="--",label="cube")
plt.xlabel("x-axis")
plt.grid()
plt.ylabel("y-axis")
plt.legend()
plt.title("Line plot")
```
