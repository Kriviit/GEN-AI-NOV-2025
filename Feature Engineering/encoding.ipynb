## when to use Encoding
- One-Hot: Low/medium cardinality; tree models and linear models.
- Label: Ordinal categories or tree-based models (not linear without caution).
- Target: High-cardinality with strong target signal; risk of leakage.
- Count/Frequency: High-cardinality; compact representation; preserves prevalence.
- Binary: Very high-cardinality; reduces dimensionality vs one-hot.

```python
# Setup: sample data
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'city': ['NY','SF','LA','LA','NY','TX','TX','SF','SF','LA'],
    'device': ['mobile','desktop','tablet','mobile','desktop','mobile','tablet','desktop','mobile','tablet'],
    'tier': ['basic','pro','basic','plus','pro','basic','plus','basic','plus','pro'],
    'y': [1,0,1,0,1,0,1,0,0,1]
})
df
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder()
```

```python
ohe.fit_transform(df[['city','device']])
```

```python
pd.DataFrame(ohe.fit_transform(df[['city','device','tier']]),columns=ohe.get_feature_names_out(['city','device','tier']))
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python

df = pd.DataFrame({
    'city': ['NY','SF','LA','LA','NY','TX','TX','SF','SF','LA'],
    'device': ['mobile','desktop','tablet','mobile','desktop','mobile','tablet','desktop','mobile','tablet'],
    'tier': ['basic','pro','basic','plus','pro','basic','plus','basic','plus','pro'],
    'y': [1,0,1,0,1,0,1,0,0,1]
})
df
```

```python
le=LabelEncoder()
df['city_encoded']=le.fit_transform(df.city)
```

```python
df[['city','city_encoded']]
```
