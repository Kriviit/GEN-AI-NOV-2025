```python
df=pd.read_csv('data.csv')
```

```python
df_copy=df.copy()
```

```python
df_copy.session_duration.unique()
```

```python
func=lambda x: '0hr '+x if 'hr' not in x else x
df_copy['session_duration']=df_copy['session_duration'].map(func)
```

```python
df_copy['session_duration'].unique()
```

```python
df_copy['session_duration']=(df_copy['session_duration'].str.split(" ").str[0].str[:-2].astype(int)*60) +(df_copy['session_duration'].str.split(" ").str[1].str[:-3].astype(int))
```

```python
df_copy.rename(columns={'session_duration':'session_duration_mins'})
```

```python
df_copy.video_views=df_copy.video_views.str[:-1].astype(int)
```
